```python
from ultralytics import YOLO
import cv2
import numpy as np
import os

print("Libraries Imported Successfully!")
```

```python
model = YOLO("yolov8n-pose.pt")

print("YOLO Pose Model Loaded Successfully!")
```

```python
from google.colab import files

uploaded = files.upload()

input_video = next(iter(uploaded.keys()))

print("Video Uploaded Successfully:", input_video)
```

```python
cap = cv2.VideoCapture(input_video)

if not cap.isOpened():
    raise Exception("Could not open input video!")

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

if fps <= 0:
    fps = 30

print("Video Loaded Successfully!")
print("Width:", width)
print("Height:", height)
print("FPS:", fps)
```

```python
output_video = "pose_video.mp4"

fourcc = cv2.VideoWriter_fourcc(*"mp4v")

out = cv2.VideoWriter(
    output_video,
    fourcc,
    fps,
    (width, height)
)

if not out.isOpened():
    raise Exception("Could not create output video!")

print("Output Video Initialized Successfully!")
```

```python
frame_count = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # Detect human pose
    results = model(
        frame,
        conf=0.25,
        verbose=False
    )

    # Create black background
    black_background = np.zeros_like(frame)

    # Draw detected human skeleton
    annotated_frame = results[0].plot(
        img=black_background,
        boxes=False,
        labels=False
    )

    # Save processed frame
    out.write(annotated_frame)

    frame_count += 1

    if frame_count % 30 == 0:
        print("Processed frames:", frame_count)

cap.release()
out.release()

print("Dance Pose Video Created Successfully!")
print("Total Frames Processed:", frame_count)
```
